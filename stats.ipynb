```python
# 3d plot of 2 variable function

def function(x, y):
    return x**2 + y**2

x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

X, Y = np.meshgrid(x, y)
Z = function(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define your trajectory
t = np.linspace(0, 10, 100)
x = np.cos(t)
y = np.sin(t)
z = t

# Create a new figure
fig = plt.figure()

# Create a 3D axis
ax = plt.axes(projection='3d')

# Plot the trajectory
ax.plot3D(x, y, z, 'gray')

# Set labels
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

# Show the plot
plt.show()
```
